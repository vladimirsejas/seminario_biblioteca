## Introduction

PySUS is a Python library that provides easy access to publicly available datasets from Brazil's Unified Health System (SUS), published by DATASUS.

It handles file discovery, downloading and parsing so users can focus on data analysis rather than dealing with legacy file formats.

This notebook presents a complete beginner-friendly workflow, from installation to the first data exploration using real SUS data.

## Installation

Install PySUS using pip.

Open a terminal (PowerShell on Windows, Terminal on Linux or macOS) and run:

pip install pysus

If you are working inside a virtual environment (recommended), activate it first.
